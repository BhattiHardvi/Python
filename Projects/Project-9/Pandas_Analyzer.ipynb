```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


class SalesDataAnalyzer:

    def __init__(self, file_path=None):
        self.data = None

        if file_path:
            self.load_data(file_path)

    def __del__(self):
        pass

    def load_data(self, file_path):
        try:
            self.data = pd.read_csv(file_path)
            print("\nDataset loaded successfully!")

        except FileNotFoundError:
            print("\nFile not found.")

        except Exception as e:
            print("\nError:", e)

    def explore_data(self):

        if self.data is None:
            print("\nPlease load dataset first.")
            return

        while True:

            print("\n========== Explore Data ==========")
            print("1. Display first 5 rows")
            print("2. Display last 5 rows")
            print("3. Display column names")
            print("4. Display data types")
            print("5. Display data information")
            print("6. Go Back")

            choice = input("Enter your choice: ")

            if choice == "1":
                print(self.data.head())

            elif choice == "2":
                print(self.data.tail())

            elif choice == "3":
                print("\nColumn Names:")
                print(list(self.data.columns))

            elif choice == "4":
                print("\nData Types:")
                print(self.data.dtypes)

            elif choice == "5":
                print("\nData Information:")
                self.data.info()

            elif choice == "6":
                break

            else:
                print("Invalid choice.")

    def dataframe_operations(self):

        if self.data is None:
            print("\nPlease load dataset first.")
            return

        while True:

            print("\n========== DataFrame Operations ==========")
            print("1. Add New Column")
            print("2. Delete Column")
            print("3. Rename Column")
            print("4. Sort Data")
            print("5. Select Column")
            print("6. Go Back")

            choice = input("Enter your choice: ")

            if choice == "1":

                column = input("Enter new column name: ")
                value = input("Enter value for new column: ")

                self.data[column] = value

                print("\nColumn added successfully!")

            elif choice == "2":

                column = input("Enter column name to delete: ")

                if column in self.data.columns:

                    self.data.drop(
                        column,
                        axis=1,
                        inplace=True
                    )

                    print("\nColumn deleted successfully!")

                else:
                    print("Column not found.")

            elif choice == "3":

                old_name = input("Enter old column name: ")
                new_name = input("Enter new column name: ")

                if old_name in self.data.columns:

                    self.data.rename(
                        columns={old_name: new_name},
                        inplace=True
                    )

                    print("\nColumn renamed successfully!")

                else:
                    print("Column not found.")

            elif choice == "4":

                column = input("Enter column name for sorting: ")

                if column in self.data.columns:

                    self.data.sort_values(
                        by=column,
                        inplace=True
                    )

                    print("\nData sorted successfully!")

                else:
                    print("Column not found.")

            elif choice == "5":

                column = input("Enter column name: ")

                if column in self.data.columns:
                    print(self.data[column])

                else:
                    print("Column not found.")

            elif choice == "6":
                break

            else:
                print("Invalid choice.")

    def clean_data(self):

        if self.data is None:
            print("\nPlease load dataset first.")
            return

        while True:

            print("\n========== Handle Missing Data ==========")
            print("1. Display rows with missing values")
            print("2. Fill missing values with mean")
            print("3. Drop rows with missing values")
            print("4. Replace missing values with specific value")
            print("5. Go Back")

            choice = input("Enter your choice: ")

            if choice == "1":

                missing = self.data[
                    self.data.isnull().any(axis=1)
                ]

                if missing.empty:
                    print(
                        "\nNo missing values found in the dataset!"
                    )
                else:
                    print(missing)

            elif choice == "2":

                numeric_columns = self.data.select_dtypes(
                    include=np.number
                ).columns

                for column in numeric_columns:

                    self.data[column] = self.data[column].fillna(
                        self.data[column].mean()
                    )

                print(
                    "\nMissing numerical values filled with mean."
                )

            elif choice == "3":

                self.data.dropna(inplace=True)

                print(
                    "\nRows with missing values removed."
                )

            elif choice == "4":

                column = input("Enter column name: ")

                if column in self.data.columns:

                    value = input(
                        "Enter replacement value: "
                    )

                    self.data[column] = (
                        self.data[column].fillna(value)
                    )

                    print(
                        "\nMissing values replaced successfully."
                    )

                else:
                    print("Column not found.")

            elif choice == "5":
                break

            else:
                print("Invalid choice.")

    def visualize_data(self):

        if self.data is None:
            print("\nPlease load dataset first.")
            return

        while True:

            print("\n========== Data Visualization ==========")
            print("1. Bar Plot")
            print("2. Line Plot")
            print("3. Scatter Plot")
            print("4. Pie Chart")
            print("5. Histogram")
            print("6. Stack Plot")
            print("7. Exit")

            choice = input("Enter your choice: ")

            if choice == "1":
                self.bar_plot()

            elif choice == "2":
                self.line_plot()

            elif choice == "3":
                self.scatter_plot()

            elif choice == "4":
                self.pie_chart()

            elif choice == "5":
                self.histogram()

            elif choice == "6":
                self.stack_plot()

            elif choice == "7":
                break

            else:
                print("Invalid choice.")

    def bar_plot(self):

        columns = list(self.data.columns)

        print("\nColumns:")
        print(columns)

        x = input("Enter x-axis column: ")
        y = input("Enter y-axis column: ")

        if x not in columns or y not in columns:
            print("Invalid column.")
            return

        plt.figure(figsize=(8, 5))

        plt.bar(
            self.data[x].astype(str),
            self.data[y]
        )

        plt.xlabel(x)
        plt.ylabel(y)
        plt.title("Bar Plot")

        plt.xticks(rotation=45)

        plt.tight_layout()
        plt.show()

        print("Bar plot displayed successfully!")

    def line_plot(self):

        columns = list(self.data.columns)

        print("\nColumns:")
        print(columns)

        x = input("Enter x-axis column: ")
        y = input("Enter y-axis column: ")

        if x not in columns or y not in columns:
            print("Invalid column.")
            return

        plt.figure(figsize=(8, 5))

        plt.plot(
            self.data[x],
            self.data[y],
            marker="o"
        )

        plt.xlabel(x)
        plt.ylabel(y)
        plt.title("Line Plot")

        plt.tight_layout()
        plt.show()

        print("Line plot displayed successfully!")

    def scatter_plot(self):

        columns = list(self.data.columns)

        print("\nColumns:")
        print(columns)

        x = input("Enter x-axis column: ")
        y = input("Enter y-axis column: ")

        if x not in columns or y not in columns:
            print("Invalid column.")
            return

        plt.figure(figsize=(8, 5))

        plt.scatter(
            self.data[x],
            self.data[y]
        )

        plt.xlabel(x)
        plt.ylabel(y)
        plt.title("Scatter Plot")

        plt.tight_layout()
        plt.show()

        print("Scatter plot displayed successfully!")

    def pie_chart(self):

        columns = list(self.data.columns)

        print("\nColumns:")
        print(columns)

        column = input(
            "Enter column for pie chart: "
        )

        if column not in columns:
            print("Invalid column.")
            return

        values = self.data[column].value_counts()

        plt.figure(figsize=(7, 7))

        plt.pie(
            values.values,
            labels=values.index,
            autopct="%1.1f%%"
        )

        plt.title("Pie Chart")

        plt.show()

        print("Pie chart displayed successfully!")

    def histogram(self):

        numeric_columns = list(
            self.data.select_dtypes(
                include=np.number
            ).columns
        )

        print("\nNumerical Columns:")
        print(numeric_columns)

        column = input(
            "Enter numerical column: "
        )

        if column not in numeric_columns:
            print("Invalid column.")
            return

        plt.figure(figsize=(8, 5))

        plt.hist(
            self.data[column].dropna(),
            bins=10
        )

        plt.xlabel(column)
        plt.ylabel("Frequency")
        plt.title("Histogram")

        plt.tight_layout()
        plt.show()

        print("Histogram displayed successfully!")

    def stack_plot(self):

        numeric_columns = list(
            self.data.select_dtypes(
                include=np.number
            ).columns
        )

        if len(numeric_columns) < 2:

            print(
                "At least two numerical columns are required."
            )

            return

        print("\nNumerical Columns:")
        print(numeric_columns)

        column1 = input("Enter first column: ")
        column2 = input("Enter second column: ")

        if (
            column1 not in numeric_columns
            or column2 not in numeric_columns
        ):

            print("Invalid column.")
            return

        x = np.arange(len(self.data))

        plt.figure(figsize=(8, 5))

        plt.stackplot(
            x,
            self.data[column1],
            self.data[column2],
            labels=[column1, column2]
        )

        plt.xlabel("Index")
        plt.ylabel("Values")
        plt.title("Stack Plot")

        plt.legend()

        plt.tight_layout()
        plt.show()

        print("Stack plot displayed successfully!")

    def save_visualization(self):

        print("\n========== Save Visualization ==========")

        file_name = input(
            "Enter file name to save the plot "
            "(e.g., plot.png): "
        )

        if not file_name.endswith(
            (".png", ".jpg", ".jpeg")
        ):
            file_name += ".png"

        plt.savefig(
            file_name,
            dpi=300,
            bbox_inches="tight"
        )

        print(
            "Visualization saved as",
            file_name,
            "successfully!"
        )


def main():

    analyzer = SalesDataAnalyzer()

    while True:

        print("\n")
        print(
            "========== Data Analysis & "
            "Visualization Program =========="
        )

        print("Please select an option:")
        print("1. Load Dataset")
        print("2. Explore Data")
        print("3. Perform DataFrame Operations")
        print("4. Handle Missing Data")
        print("5. Data Visualization")
        print("6. Save Visualization")
        print("7. Exit")

        print(
            "=========================================================="
        )

        choice = input("Enter your choice: ")

        if choice == "1":

            file_path = input(
                "\nEnter the path of the dataset (CSV file): "
            )

            analyzer.load_data(file_path)

        elif choice == "2":

            analyzer.explore_data()

        elif choice == "3":

            analyzer.dataframe_operations()

        elif choice == "4":

            analyzer.clean_data()

        elif choice == "5":

            analyzer.visualize_data()

        elif choice == "6":

            analyzer.save_visualization()

        elif choice == "7":

            print("\nExiting the program. Goodbye!")
            break

        else:

            print("\nInvalid choice. Please try again.")


if __name__ == "__main__":
    main()
```
