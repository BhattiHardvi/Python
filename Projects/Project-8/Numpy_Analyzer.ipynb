```python
import numpy as np

class DataAnalytics:

    def __init__(self):
        self.array = None

    @classmethod
    def welcome(cls):
        print("\nWelcome to the NumPy Analyzer!")
        print("================================")
    @staticmethod
    def show_message(message):
        print(message)

    def create_array(self):
        print("\nSelect the type of array to create:")
        print("1. 1D Array")
        print("2. 2D Array")
        print("3. 3D Array")

        choice = int(input("Enter your choice: "))

        if choice == 1:
            values = list(map(int, input("Enter elements separated by space: ").split()))
            self.array = np.array(values)
            
        elif choice == 2:
            rows = int(input("Enter the number of rows: "))
            cols = int(input("Enter the number of columns: "))
            values = list(map(int, input(f"Enter {rows * cols} elements separated by space: ").split()))

            if len(values) != rows * cols:
                print("Invalid number of elements.")
                return

            self.array = np.array(values).reshape(rows, cols)

        elif choice == 3:
            d1 = int(input("Enter depth: "))
            rows = int(input("Enter number of rows: "))
            cols = int(input("Enter number of columns: "))

            values = list(map(float, input(f"Enter {d1 * rows * cols} elements separated by space: ").split()))

            if len(values) != d1 * rows * cols:
                print("Invalid number of elements.")
                return
                
            self.array = np.array(values).reshape(d1, rows, cols)

        else:
            print("Invalid choice.")
            return

        print("\nArray created successfully:")
        print(self.array)

    def indexing_slicing(self):
        if self.array is None:
            print("Please create an array first.")
            return

        while True:
            print("\nChoose an operation:")
            print("1. Indexing")
            print("2. Slicing")
            print("3. Go Back")

            choice = int(input("Enter your choice: "))

            if choice == 1:
                if self.array.ndim == 1:
                    index = int(input("Enter index: "))
                    print("Element:", self.array[index])

                else:
                    row = int(input("Enter row index: "))
                    col = int(input("Enter column index: "))
                    print("Element:", self.array[row, col])

            elif choice == 2:
                if self.array.ndim == 1:
                    start = int(input("Enter start index: "))
                    end = int(input("Enter end index: "))
                    print("\nSliced Array:")
                    print(self.array[start:end])

                else:
                    row_range = input("Enter row range (start:end): ")
                    col_range = input("Enter column range (start:end): ")

                    rs, re = map(int, row_range.split(":"))
                    cs, ce = map(int, col_range.split(":"))

                    print("\nSliced Array:")
                    print(self.array[rs:re, cs:ce])

            elif choice == 3:
                break

            else:
                print("Invalid choice.")

    def mathematical_operations(self):
        if self.array is None:
            print("Please create an array first.")
            return

        print("\nMathematical Operations:")
        print("1. Addition")
        print("2. Subtraction")
        print("3. Multiplication")
        print("4. Division")
        

        choice = int(input("Enter your choice: "))

        if choice in [1, 2, 3, 4]:
            if self.array.ndim == 1:
                size = self.array.size
            else:
                size = self.array.size

            values = list(map(int, input(f"Enter the same-size array elements ({size} elements separated by space): ").split()))

            if len(values) != size:
                print("Invalid number of elements.")
                return

            second = np.array(values).reshape(self.array.shape)

            print("\nOriginal Array:")
            print(self.array)

            print("\nSecond Array:")
            print(second)

            if choice == 1:
                result = self.array + second
                name = "Addition"

            elif choice == 2:
                result = self.array - second
                name = "Subtraction"

            elif choice == 3:
                result = self.array * second
                name = "Multiplication"

            else:
                if np.any(second == 0):
                    print("Division by zero is not allowed.")
                    return

                result = self.array / second
                name = "Division"

            print(f"\nResult of {name}:")
            print(result)
        
        elif choice == 5:
            values = list(map(int, input(f"Enter {self.array.size} elements for second array: ").split()))

            if len(values) != self.array.size:
                print("Invalid number of elements.")
                return

            second = np.array(values).reshape(self.array.shape)

            print("\nDot Product:")
            print(np.dot(self.array.flatten(), second.flatten()))

        elif choice == 6:
            if self.array.ndim != 2:
                print("Matrix multiplication requires a 2D array.")
                return

            rows = int(input("Enter rows of second matrix: "))
            cols = int(input("Enter columns of second matrix: "))

            if self.array.shape[1] != rows:
                print("Invalid dimensions. Columns of first matrix ,must equal rows of second matrix.")
                return

            values = list(map(int, input(f"Enter {rows * cols} elements: ").split()))

            if len(values) != rows * cols:
                print("Invalid number of elements.")
                return

            second = np.array(values).reshape(rows, cols)

            print("\nFirst Matrix:")
            print(self.array)

            print("\nSecond Matrix:")
            print(second)

            print("\nMatrix Multiplication:")
            print(np.matmul(self.array, second))

        else:
            print("Invalid choice.")
            
    def combine_split(self):
        if self.array is None:
            print("Please create an array first.")
            return

        print("\nCombine or Split Arrays:")
        print("1. Combine Arrays")
        print("2. Split Array")

        choice = int(input("Enter your choice: "))

        if choice == 1:
            if self.array.ndim != 2:
                print("Combine operation requires a 2D array.")
                return

            values = list(map(int, input(f"Enter the elements of another array to combine "f"({self.array.size} elements separated by space): ").split()))

            if len(values) != self.array.size:
                print("Invalid number of elements.")
                return

            second = np.array(values).reshape(self.array.shape)

            print("\nOriginal Array:")
            print(self.array)

            print("\nSecond Array:")
            print(second)

            print("\nCombined Array (Vertical Stack):")
            print(np.vstack((self.array, second)))

        elif choice == 2:
            sections = int(input("Enter number of sections: "))

            try:
                result = np.array_split(self.array, sections)
                print("\nSplit Arrays:")
                for i, part in enumerate(result, 1):
                    print(f"Array {i}:")
                    print(part)
            except:
                print("Array cannot be split into these sections.")

        else:
            print("Invalid choice.")

    def search_sort_filter(self):
        if self.array is None:
            print("Please create an array first.")
            return

        print("\nSearch, Sort, and Filter:")
        print("1. Search a value")
        print("2. Sort the array")
        print("3. Filter values")

        choice = int(input("Enter your choice: "))

        if choice == 1:
            value = float(input("Enter value to search: "))
            result = np.where(self.array == value)

            print("\nSearch Result:")
            print(result)

        elif choice == 2:
            print("\nOriginal Array:")
            print(self.array)

            print("\n1. Ascending")
            print("2. Descending")

            order = int(input("Enter sorting order: "))

            if self.array.ndim == 1:
                sorted_array = np.sort(self.array)

                if order == 2:
                    sorted_array = sorted_array[::-1]
            else:
                sorted_array = np.sort(self.array, axis=1)

                if order == 2:
                    sorted_array = np.flip(sorted_array, axis=1)

            print("\nSorted Array:")
            print(sorted_array)

        elif choice == 3:
            value = float(input("Filter values greater than: "))

            print("\nFiltered Values:")
            print(self.array[self.array > value])

        else:
            print("Invalid choice.")

    def aggregates_statistics(self):
        if self.array is None:
            print("Please create an array first.")
            return

        print("\nAggregates and Statistics:")
        print("1. Sum")
        print("2. Mean")
        print("3. Median")
        print("4. Standard Deviation")
        print("5. Variance")
        

        choice = int(input("Enter your choice: "))

        print("\nOriginal Array:")
        print(self.array)

        if choice == 1:
            print("\nSum of Array:", np.sum(self.array))

        elif choice == 2:
            print("\nMean of Array:", np.mean(self.array))

        elif choice == 3:
            print("\nMedian of Array:", np.median(self.array))

        elif choice == 4:
            print("\nStandard Deviation:", np.std(self.array))

        elif choice == 5:
            print("\nVariance:", np.var(self.array))

        else:
            print("Invalid choice.")


def main():

    DataAnalytics.welcome()

    analyzer = DataAnalytics()

    while True:

        print("Choose an option:")
        print("1. Create a Numpy Array")
        print("2. Perform Mathematical Operations")
        print("3. Combine or Split Arrays")
        print("4. Search, Sort, or Filter Arrays")
        print("5. Compute Aggregates and Statistics")
        print("6. Exit")

        choice = int(input("Enter your choice: "))

        if choice == 1:
            analyzer.create_array()
            analyzer.indexing_slicing()

        elif choice == 2:
            analyzer.mathematical_operations()

        elif choice == 3:
            analyzer.combine_split()

        elif choice == 4:
            analyzer.search_sort_filter()

        elif choice == 5:
            analyzer.aggregates_statistics()

        elif choice == 6:
            print("\nThank you for using the NumPy Analyzer! Goodbye!")
            break

        else:
            print("Invalid choice. Please try again.")


main()
```
